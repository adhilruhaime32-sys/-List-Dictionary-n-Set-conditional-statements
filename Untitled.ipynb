```python
age_list = [24, 25, 26, 27, 28]
name_list = ["Arun", "Bala", "Charan", "Divya", "Esha"]
```

```python
name_list.append("Yazhini")
age_list.insert(2, 30)
name_list.remove("Yazhini")
age_list.pop()
age_list.extend([29, 30, 26])
age_list.sort(reverse=True)
print(age_list)

print("Max:", max(age_list))
print("Min:", min(age_list))
print("Sum:", sum(age_list))
```

```python
my_set = set(['a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'])
print(my_set)
```

A set stores only unique elements, so the duplicate 'a' and 'i' values are removed. Sets are also unordered, so the output order may differ from the input and can change between runs.

```python
my_set[4] = 's'
```

Sets are unordered, so elements have no index or position. That means indexing and item assignment aren't supported. To add an element, use my_set.add('s')

```python
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}
print("Union:", set1 | set2)
print("Intersection:", set1 & set2)
```
